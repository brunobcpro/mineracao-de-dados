```python
def create_database(db_path):
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    cursor.execute('''CREATE TABLE IF NOT EXISTS relatorio (
                        numero_relatorio TEXT PRIMARY KEY,
                        data TEXT
                        )''')

    for topico in ['A', 'B', 'C', 'D']:
        cursor.execute(f'''CREATE TABLE IF NOT EXISTS topico_{topico} (
                            id INTEGER PRIMARY KEY AUTOINCREMENT,
                            numero_relatorio TEXT,
                            paragrafo TEXT,
                            FOREIGN KEY (numero_relatorio) REFERENCES relatorio(numero_relatorio))
                            ''')

    conn.commit()
    conn.close()
```

```python
def extrair_dados_arquivo(nome_arquivo):
    """
    Extrai a data e o número do relatório a partir do nome do arquivo.
    Exemplo de nome de arquivo: 2020_09_COPOM233 -> data: 2020_09, número: 233
    """
    match = re.search(r'(\d{4}_\d{2})_COPOM(\d+)', nome_arquivo)
    if match:
        data = match.group(1)
        numero_relatorio = match.group(2)
        return numero_relatorio, data
    return None, None
```

```python
def processar_arquivos(diretorio, db_path):
    """
    Lê todos os arquivos do diretório, extrai as informações e insere na tabela relatorio.
    """
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    for arquivo in os.listdir(diretorio):
        numero_relatorio, data = extrair_dados_arquivo(arquivo)

        if numero_relatorio and data:
            try:
                cursor.execute('''INSERT INTO relatorio (numero_relatorio, data)
                                  VALUES (?, ?)''', (numero_relatorio, data))
            except sqlite3.IntegrityError:
                print(f"Relatório {numero_relatorio} já existe no banco de dados.")

    conn.commit()
    conn.close()
```

```python
import os
import re
import fitz  # PyMuPDF
import pdfplumber

import re
from collections import defaultdict

import nltk
from nltk.corpus import stopwords
import string

from bertopic import BERTopic
from sentence_transformers import SentenceTransformer

```

```python
"""
    Lê todos os arquivos PDF do diretório e extrai o texto, armazenando
    em um dicionário onde a chave é o número do relatório.
    Se PyMuPDF falhar, tenta com pdfplumber.
    Exibe quais arquivos foram processados com sucesso ou falharam.
"""
def extrair_texto_pdfs(diretorio):
    textos_relatorios = {}
    sucesso = 0  # Contador de arquivos extraídos com sucesso
    erro = 0     # Contador de arquivos com falha
    erros_detectados = []  # Lista para armazenar os arquivos com erro

    for arquivo in os.listdir(diretorio):
        numero_relatorio, _ = extrair_dados_arquivo(arquivo)  # Extrai número

        if numero_relatorio:
            caminho_pdf = os.path.join(diretorio, arquivo)
            texto_final = None  # Armazena o texto extraído

            # Primeiro, tenta com PyMuPDF
            try:
                with fitz.open(caminho_pdf) as pdf:
                    texto_extraido = [pagina.get_text("text") for pagina in pdf]
                    texto_final = "\n".join(texto_extraido)

                if not texto_final.strip():  # Se o texto estiver vazio, tenta outro método
                    raise ValueError("Texto vazio extraído com PyMuPDF")

            except Exception as e:
                print(f"⚠️ Falha com PyMuPDF no arquivo {arquivo}: {e}")

                # Se PyMuPDF falhar, tenta pdfplumber
                try:
                    with pdfplumber.open(caminho_pdf) as pdf:
                        texto_extraido = [pagina.extract_text() for pagina in pdf.pages if pagina.extract_text()]
                        texto_final = "\n".join(texto_extraido)

                    if not texto_final.strip():  # Verifica se ainda está vazio
                        raise ValueError("Texto vazio extraído com pdfplumber")

                except Exception as e2:
                    erro += 1
                    erros_detectados.append((arquivo, str(e2)))
                    print(f"❌ Erro ao processar {arquivo} com pdfplumber: {e2}")
                    continue  # Pula para o próximo arquivo

            # Se chegou até aqui, conseguiu extrair o texto corretamente
            textos_relatorios[numero_relatorio] = texto_final
            sucesso += 1
            print(f"✅ Extraído com sucesso: {arquivo}")

            # Detecta possível texto em múltiplas colunas
            if "\n\n" in texto_final[:500]:  # Olha apenas no começo do texto
                print(f"🔍 O arquivo {arquivo} pode ter múltiplas colunas.")

    # Exibe resumo final
    print(f"\n📊 Resumo: {sucesso} arquivos extraídos com sucesso, {erro} arquivos com erro.\n")

    # Lista de arquivos com erro
    if erros_detectados:
        print("❌ Arquivos com erro:")
        for arq, motivo in erros_detectados:
            print(f"- {arq}: {motivo}")

    return textos_relatorios
```

```python
def extrair_dados_arquivo(nome_arquivo):
    """
    Extrai o número do relatório do nome do arquivo.
    Exemplo: '2020_09_COPOM233.pdf' -> numero: 233
    """
    match = re.search(r'(\d{4}_\d{2})_COPOM(\d+)', nome_arquivo)
    if match:
        return match.group(2), match.group(1)  # (numero_relatorio, data)
    return None, None
```

```python
def dividir_texto_em_partes(textos_relatorios, frases_busca):
    """
    Divide o texto de cada relatório em 4 partes com base nas frases encontradas.

    Parâmetros:
    - textos_relatorios: dicionário {numero_relatorio: texto}
    - frases_busca: lista de frases que definem as divisões

    Retorna:
    - Dicionário {numero_relatorio: {parte_1: "texto", parte_2: "texto", ...}}
    """
    resultados_divididos = {}

    for numero_relatorio, texto in textos_relatorios.items():
        # Encontra as posições das frases no texto
        posicoes = []
        for frase in frases_busca:
            match = re.search(re.escape(frase), texto)
            if match:
                posicoes.append((match.start(), frase))

        # Ordena pela posição no texto
        posicoes.sort()

        if not posicoes:
            continue  # Se nenhuma frase foi encontrada, pula o relatório

        # Cria as divisões do texto
        partes = {}
        total_frases = len(posicoes)

        for i in range(4):
            if i < total_frases:
                inicio = posicoes[i][0]
                fim = posicoes[i + 1][0] if i + 1 < total_frases else len(texto)
                partes[f"parte_{i+1}"] = texto[inicio:fim].strip()
            else:
                partes[f"parte_{i+1}"] = ""  # Caso não tenha 4 partes, preenche com vazio

        resultados_divididos[numero_relatorio] = partes

        print(f"📄 Relatório {numero_relatorio} dividido em 4 partes!")

    return resultados_divididos
```

```python
def criar_dataframe_secoes(textos_divididos):
    """
    Cria um DataFrame separando os textos por seções, com base nos números do tipo '24.'

    Parâmetros:
    - textos_divididos: dicionário {numero_relatorio: {parte_1: "texto", parte_2: "texto", ...}}

    Retorna:
    - DataFrame com as colunas ["texto", "parte", "relatorio"]
    """
    dados = []

    for numero_relatorio, partes in textos_divididos.items():
        for parte, texto in partes.items():
            # Divide o texto pelas seções que começam com números seguidos de ponto (ex: "24.")
            secoes = re.split(r'\n(\d+\.)', texto)

            if len(secoes) > 1:
                # Reconstruir as seções corretamente
                secoes_formatadas = [secoes[i] + secoes[i+1] for i in range(1, len(secoes)-1, 2)]
            else:
                secoes_formatadas = [texto]  # Caso não tenha seções, mantém o texto inteiro

            # Adicionar ao DataFrame
            for secao in secoes_formatadas:
                dados.append({
                    "texto": secao.strip(),
                    "parte": parte,
                    "relatorio": numero_relatorio
                })

    df = pd.DataFrame(dados, columns=["texto", "parte", "relatorio"])

    return df

```

```python
def remover_stopwords(texto):
    """
    Remove as stopwords e pontuações do texto.
    """
    stop_words = set(stopwords.words('portuguese'))  # Stopwords em português
    palavras = texto.split()  # Quebra o texto em palavras
    palavras_filtradas = [palavra for palavra in palavras if palavra.lower() not in stop_words]  # Remove stop-words
    texto_limpo = " ".join(palavras_filtradas)  # Junta novamente as palavras
    return texto_limpo
```

```python
# Modelo de embeddings mais moderno
modelo_transformer = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
"""
    Aplica BERTopic para extração de tópicos em um DataFrame de textos.

    Parâmetros:
    - df_textos: DataFrame com a coluna 'texto'.
    - nome_parte: Nome da parte para referência.

    Retorna:
    - df_topicos: DataFrame com os tópicos extraídos.
    """

def extrair_topicos(df_textos, nome_parte, n_palavras=5):

    if df_textos.empty:
        print(f"⚠️ Nenhum texto encontrado para {nome_parte}.")
        return df_textos.assign(topico=None, palavras_topico=None)

    textos = df_textos["texto"].tolist()  # Coletar textos

    # Criar e ajustar modelo BERTopic
    modelo = BERTopic(embedding_model=modelo_transformer, verbose=True)
    topicos, _ = modelo.fit_transform(textos)

    # Criar DataFrame com os resultados
    df_topicos = df_textos.copy()
    df_topicos["topico"] = topicos

    # Coletar palavras-chave dos tópicos
    topicos_info = modelo.get_topic_info().set_index("Topic")["Representation"].to_dict()
    df_topicos["palavras_topico"] = df_topicos["topico"].map(lambda t: ", ".join(topicos_info.get(t, [])[:n_palavras]))

    print(f"✅ Extração de tópicos concluída para {nome_parte}!")

    return df_topicos
```

```python
import sqlite3
import pandas as pd
from bertopic import BERTopic

def extrair_topicos_bd(nome_banco, modelo_transformer, n_palavras=5):
    # Conectar ao banco
    conn = sqlite3.connect(nome_banco)

    # Listar as tabelas a serem processadas
    tabelas = {
        "parte1_dados": "Parte 1",
        "parte2_dados": "Parte 2",
        "parte3_dados": "Parte 3",
        "parte4_dados": "Parte 4"
    }

    resultados = []

    for tabela, parte in tabelas.items():
        print(f"📌 Processando {tabela}...")

        # Carregar dados da tabela
        df = pd.read_sql(f"SELECT texto, texto_limpo, numero_relatorio FROM {tabela}", conn)

        if df.empty:
            print(f"⚠️ Nenhum dado encontrado em {tabela}. Pulando...")
            continue

        # Aplicar BERTopic na coluna 'texto_limpo'
        modelo = BERTopic(embedding_model=modelo_transformer, verbose=True)
        topicos, _ = modelo.fit_transform(df["texto_limpo"].tolist())

        # Obter palavras-chave dos tópicos
        topicos_info = modelo.get_topic_info().set_index("Topic")["Representation"].to_dict()
        palavras_topico = [", ".join(topicos_info.get(t, [])[:n_palavras]) for t in topicos]

        # Adicionar os resultados ao DataFrame
        df["topico"] = topicos
        df["palavras_topico"] = palavras_topico
        df["parte"] = parte  # Adicionar a informação da parte

        # Guardar no consolidado
        resultados.append(df)

        print(f"✅ Extração concluída para {tabela}!")

    conn.close()

    # Concatenar os resultados em um único DataFrame
    if resultados:
        df_final = pd.concat(resultados, ignore_index=True)
        return df_final
    else:
        return pd.DataFrame(columns=["texto", "texto_limpo", "numero_relatorio", "topico", "palavras_topico", "parte"])

# Exemplo de uso:
df_resultado = extrair_topicos_bd("relatorio.db", modelo_transformer)
df_resultado.to_csv("topicos_extraidos.csv", index=False)
```

```python
import sqlite3

def salvar_contagem_topicos(db_path, topicos_contagem_parte1, topicos_contagem_parte2, topicos_contagem_parte3, topicos_contagem_parte4):
    """
    Salva as contagens de tópicos em 4 tabelas dentro de um banco de dados SQLite.

    Parâmetros:
    - db_path: Caminho do banco de dados SQLite.
    - topicos_contagem_parte1, topicos_contagem_parte2, topicos_contagem_parte3, topicos_contagem_parte4: DataFrames com a contagem de tópicos.
    """
    # Conectar ao banco de dados SQLite
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    # Criar tabelas para armazenar as contagens de tópicos se não existirem
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS parte1_topicos (
            topico TEXT,
            contagem INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE IF NOT EXISTS parte2_topicos (
            topico TEXT,
            contagem INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE IF NOT EXISTS parte3_topicos (
            topico TEXT,
            contagem INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE IF NOT EXISTS parte4_topicos (
            topico TEXT,
            contagem INTEGER
        )
    ''')

    # Função para inserir os dados na tabela correspondente
    def salvar_topicos(df, parte):
        for _, row in df.iterrows():
            cursor.execute(f'''
                INSERT INTO {parte}_topicos (topico, contagem)
                VALUES (?, ?)
            ''', (row['topico'], row['contagem']))

    # Salvar as contagens para cada parte
    salvar_topicos(topicos_contagem_parte1, 'parte1')
    salvar_topicos(topicos_contagem_parte2, 'parte2')
    salvar_topicos(topicos_contagem_parte3, 'parte3')
    salvar_topicos(topicos_contagem_parte4, 'parte4')

    # Commit das alterações e fechamento da conexão
    conn.commit()
    conn.close()

# Exemplo de uso
salvar_contagem_topicos('relatorio.db', contagem_topicos_parte1, contagem_topicos_parte2, contagem_topicos_parte3, contagem_topicos_parte4)
```

```python
import sqlite3

def salvar_dados_completos_em_tabelas(db_path, df_parte1, df_parte2, df_parte3, df_parte4):
    """
    Salva os dados completos (tópico, texto, parte, relatório) em 4 tabelas dentro de um banco de dados SQLite.

    Parâmetros:
    - db_path: Caminho do banco de dados SQLite.
    - df_parte1, df_parte2, df_parte3, df_parte4: DataFrames com os dados completos (tópico, texto, parte, relatório).
    """
    # Verificar as colunas dos DataFrames
    print("Colunas do DataFrame Parte 1:", df_parte1.columns)
    print("Colunas do DataFrame Parte 2:", df_parte2.columns)
    print("Colunas do DataFrame Parte 3:", df_parte3.columns)
    print("Colunas do DataFrame Parte 4:", df_parte4.columns)

    # Conectar ao banco de dados SQLite
    conn = sqlite3.connect(db_path)
    cursor = conn.cursor()

    # Apagar tabelas existentes, caso existam
    cursor.execute('DROP TABLE IF EXISTS parte1_dados')
    cursor.execute('DROP TABLE IF EXISTS parte2_dados')
    cursor.execute('DROP TABLE IF EXISTS parte3_dados')
    cursor.execute('DROP TABLE IF EXISTS parte4_dados')

    # Criar novas tabelas para armazenar os dados completos
    cursor.execute('''
        CREATE TABLE parte1_dados (
            topico TEXT,
            texto TEXT,
            parte INTEGER,
            numero_relatorio INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE parte2_dados (
            topico TEXT,
            texto TEXT,
            parte INTEGER,
            numero_relatorio INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE parte3_dados (
            topico TEXT,
            texto TEXT,
            parte INTEGER,
            numero_relatorio INTEGER
        )
    ''')

    cursor.execute('''
        CREATE TABLE parte4_dados (
            topico TEXT,
            texto TEXT,
            parte INTEGER,
            numero_relatorio INTEGER
        )
    ''')

    # Função para salvar os dados completos
    def salvar_dados(df, parte):
        # Verificar se o DataFrame possui as colunas corretas
        if 'topico' not in df.columns or 'texto' not in df.columns or 'relatorio' not in df.columns:
            print(f"Erro: o DataFrame da Parte {parte} está faltando uma ou mais colunas necessárias.")
            print("Colunas encontradas:", df.columns)
            return

        # Inserir os dados na tabela correspondente
        for _, row in df.iterrows():
            cursor.execute(f'''
                INSERT INTO parte{parte}_dados (topico, texto, parte, numero_relatorio)
                VALUES (?, ?, ?, ?)
            ''', (row['topico'], row['texto'], parte, row['relatorio']))

    # Salvar os dados completos para cada parte
    salvar_dados(df_parte1, 1)
    salvar_dados(df_parte2, 2)
    salvar_dados(df_parte3, 3)
    salvar_dados(df_parte4, 4)

    # Commit das alterações e fechamento da conexão
    conn.commit()
    conn.close()

# Exemplo de uso:
salvar_dados_completos_em_tabelas('relatorio.db', df_parte1, df_parte2, df_parte3, df_parte4)
```
